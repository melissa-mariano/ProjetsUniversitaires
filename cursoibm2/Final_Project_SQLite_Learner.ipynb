## Problem 1

#### Create tables

Establish a connection **conn** to the RSQLite database **FinalDB.sqlite**, and create the following four tables.

1.  **CROP_DATA**
2.  **FARM_PRICES**
3.  **DAILY_FX**
4.  **MONTHLY_FX**

The previous practice lab will help you accomplish this.


```r
# Write your query here

# Establish connection
conn <- dbConnect(RSQLite::SQLite(), "FinalDB.sqlite")

# (Opcional, caso você já tenha executado antes)
dbExecute(conn, "DROP TABLE IF EXISTS CROP_DATA")
dbExecute(conn, "DROP TABLE IF EXISTS FARM_PRICES")
dbExecute(conn, "DROP TABLE IF EXISTS DAILY_FX")
dbExecute(conn, "DROP TABLE IF EXISTS MONTHLY_FX")

# Create CROP_DATA table
dbExecute(conn, "
CREATE TABLE CROP_DATA (
    CD_ID INTEGER,
    YEAR INTEGER,
    CROP_TYPE TEXT,
    GEO TEXT,
    SEEDED_AREA REAL,
    HARVESTED_AREA REAL,
    PRODUCTION REAL,
    AVG_YIELD REAL
);
")

# Create FARM_PRICES table
dbExecute(conn, "
CREATE TABLE FARM_PRICES (
    CD_ID INTEGER,
    DATE TEXT,
    CROP_TYPE TEXT,
    GEO TEXT,
    PRICE_PRERMT REAL
);
")

# Create DAILY_FX table
dbExecute(conn, "
CREATE TABLE DAILY_FX (
    DFX_ID INTEGER,
    DATE TEXT,
    FXUSDCAD REAL
);
")

# Create MONTHLY_FX table
dbExecute(conn, "
CREATE TABLE MONTHLY_FX (
    DFX_ID INTEGER,
    DATE TEXT,
    FXUSDCAD REAL
);
")
```

```r
#check list of tables in the present db.
dbListTables(conn)
```

## Problem 2

#### Read Datasets and load your tables in database

Read the datasets into R dataframes using the urls provided above. Then load your tables in database.


```r
# Write your query here

# Dataset URLs
crop_url <- "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-RP0203EN-SkillsNetwork/labs/Final%20Project/Annual_Crop_Data.csv"
farm_url <- "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-RP0203EN-SkillsNetwork/labs/Final%20Project/Monthly_Farm_Prices.csv"
daily_fx_url <- "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-RP0203EN-SkillsNetwork/labs/Final%20Project/Daily_FX.csv"
monthly_fx_url <- "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-RP0203EN-SkillsNetwork/labs/Final%20Project/Monthly_FX.csv"

# Read datasets
crop_data <- read.csv(crop_url, stringsAsFactors = FALSE)
farm_prices <- read.csv(farm_url, stringsAsFactors = FALSE)
daily_fx <- read.csv(daily_fx_url, stringsAsFactors = FALSE)
monthly_fx <- read.csv(monthly_fx_url, stringsAsFactors = FALSE)

# Load tables
dbWriteTable(conn, "CROP_DATA", crop_data, append = TRUE, row.names = FALSE)
dbWriteTable(conn, "FARM_PRICES", farm_prices, append = TRUE, row.names = FALSE)
dbWriteTable(conn, "DAILY_FX", daily_fx, append = TRUE, row.names = FALSE)
dbWriteTable(conn, "MONTHLY_FX", monthly_fx, append = TRUE, row.names = FALSE)
```

## Now execute SQL queries using the RSQLite R package to solve the assignment problems.

## Problem 3

#### How many records are in the farm prices dataset?


```r
# Write your query here

query <- "
SELECT COUNT(*) AS TOTAL_RECORDS
FROM FARM_PRICES;
"

dbGetQuery(conn, query)
```

## Problem 4

#### Which geographies are included in the farm prices dataset?


```r
# Write your query here

query <- "
SELECT DISTINCT GEO
FROM FARM_PRICES
ORDER BY GEO;
"

dbGetQuery(conn, query)
```

## Problem 5

#### How many hectares of Rye were harvested in Canada in 1968?


```r
# Write your query here

query <- "
SELECT HARVESTED_AREA
FROM CROP_DATA
WHERE CROP_TYPE = 'Rye'
  AND GEO = 'Canada'
  AND YEAR = '1968-12-31';
"

dbGetQuery(conn, query)
```

## Problem 6

#### Query and display the first 6 rows of the farm prices table for Rye.


```r
# Wriet your query here

query <- "
SELECT *
FROM FARM_PRICES
WHERE CROP_TYPE = 'Rye'
LIMIT 6;
"

dbGetQuery(conn, query)
```

## Problem 7

#### Which provinces grew Barley?


```r
# Write your query here

query <- "
SELECT DISTINCT GEO
FROM CROP_DATA
WHERE CROP_TYPE = 'Barley'
  AND GEO <> 'Canada'
ORDER BY GEO;
"

dbGetQuery(conn, query)
```

## Problem 8

#### Find the first and last dates for the farm prices data.


```r
# Write your query here

query <- "
SELECT
    MIN(DATE) AS FIRST_DATE,
    MAX(DATE) AS LAST_DATE
FROM FARM_PRICES;
"

dbGetQuery(conn, query)
```

## Problem 9

#### Which crops have ever reached a farm price greater than or equal to $350 per metric tonne?


```r
# Write your query here

query <- "
SELECT DISTINCT CROP_TYPE
FROM FARM_PRICES
WHERE PRICE_PRERMT >= 350
ORDER BY CROP_TYPE;
"

dbGetQuery(conn, query)
```

## Problem 10

#### Rank the crop types harvested in Saskatchewan in the year 2000 by their average yield. Which crop performed best?


```r
# Write your query here

query <- "
SELECT
    CROP_TYPE,
    AVG_YIELD
FROM CROP_DATA
WHERE GEO = 'Saskatchewan'
  AND substr(YEAR,1,4) = '2000'
ORDER BY AVG_YIELD DESC;
"

dbGetQuery(conn, query)
```

## Problem 11

#### Rank the crops and geographies by their average yield (KG per hectare) since the year 2000. Which crop and province had the highest average yield since the year 2000?


```r
# Write your query here

query <- "
SELECT
    CROP_TYPE,
    GEO,
    AVG(AVG_YIELD) AS AVG_YIELD
FROM CROP_DATA
WHERE substr(YEAR,1,4) >= '2000'
  AND GEO <> 'Canada'
GROUP BY CROP_TYPE, GEO
ORDER BY AVG_YIELD DESC;
"

dbGetQuery(conn, query)
```

## Problem 12

#### Use a subquery to determine how much wheat was harvested in Canada in the most recent year of the data.


```r
# Write your query here

query <- "
SELECT HARVESTED_AREA
FROM CROP_DATA
WHERE CROP_TYPE = 'Wheat'
  AND GEO = 'Canada'
  AND YEAR = (
      SELECT MAX(YEAR)
      FROM CROP_DATA
  );
"

dbGetQuery(conn, query)
```

## Problem 13

#### Use an implicit inner join to calculate the monthly price per metric tonne of Canola grown in Saskatchewan in both Canadian and US dollars. Display the most recent 6 months of the data.


```r
# Write your query here

query <- "
SELECT
    F.DATE,
    F.PRICE_PRERMT AS PRICE_CANADIAN_DOLLARS,
    ROUND(F.PRICE_PRERMT / M.FXUSDCAD, 2) AS PRICE_US_DOLLARS
FROM FARM_PRICES F, MONTHLY_FX M
WHERE F.DATE = M.DATE
  AND F.CROP_TYPE = 'Canola'
  AND F.GEO = 'Saskatchewan'
ORDER BY F.DATE DESC
LIMIT 6;
"

dbGetQuery(conn, query)
```
